```python
# Creating a frozen set
frozen_set_example = frozenset([1, 2, 3, 4, 4])

# Trying to add or remove an element would result in an error
# frozen_set_example.add(5)  # Raises an AttributeError

# But we can use it as a key in a dictionary
my_dict = {frozen_set_example: "I am immutable!"}
print(my_dict)
```
